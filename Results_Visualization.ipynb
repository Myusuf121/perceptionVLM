```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch

# Data extracted directly from Table 4
benchmarks = [f"B{i}" for i in range(1, 31)]

clip_best = [
    'MetaVLM', '≡', '≡', 'MetaVLM', 'MetaVLM', '≡', '≡', 'MetaVLM', '≡', 'Rand',
    'MetaVLM', '≡', '≡', '≡', 'MetaVLM', 'MetaVLM', 'MetaVLM', '≡', '≡', '≡',
    'MetaVLM', '≡', 'MetaVLM', 'MetaVLM', '≡', 'MetaVLM', 'MetaVLM', 'MetaVLM', 'MetaVLM', 'MetaVLM'
]

blip_best = [
    '≡', 'Rand', '≡', 'MetaVLM', '≡', 'Rand', 'Rand', '≡', 'MetaVLM', '≡',
    '≡', 'Rand', 'MetaVLM', '≡', 'MetaVLM', 'MetaVLM', 'MetaVLM', 'MetaVLM', 'Rand', '≡',
    'MetaVLM', 'MetaVLM', '≡', 'Rand', 'Rand', '≡', 'MetaVLM', '≡', '≡', 'MetaVLM'
]

# Consistent Color Palette
color_map = {
    'MetaVLM': '#2b5c8f',  # Navy Blue
    'Rand': '#e07a5f',     # Terracotta Orange
    '≡': '#d1d5db'         # Light Gray / Slate
}

val_map = {'MetaVLM': 1, '≡': 0, 'Rand': -1}
clip_num = [val_map[x] for x in clip_best]
blip_num = [val_map[x] for x in blip_best]

# Create Figure Layout
fig = plt.figure(figsize=(15, 8.5))
gs = fig.add_gridspec(2, 1, height_ratios=[1.2, 0.8], hspace=0.45)

ax_matrix = fig.add_subplot(gs[0])
ax_summary = fig.add_subplot(gs[1])

# --- Panel 1: Categorical Benchmark Winner Matrix ---
for row_idx, (model_name, row_data, best_list) in enumerate([("CLIP", clip_num, clip_best), ("BLIP", blip_num, blip_best)]):
    for col_idx, (val, label) in enumerate(zip(row_data, best_list)):
        color = color_map[label]
        # Reduce height from 0.8 to 0.4 (or your preferred thickness)
        # Adjust the starting Y-coordinate from -0.4 to -0.2 to keep it centered on the row
        box_height = 0.45
        rect = plt.Rectangle((col_idx - 0.4, row_idx - (box_height / 2)), 0.8, box_height, color=color, ec='white', lw=1.5)
        #rect = plt.Rectangle((col_idx - 0.4, row_idx - 0.4), 0.8, 0.8, color=color, ec='white', lw=1.5)
        ax_matrix.add_patch(rect)

        text_color = 'white' if label != '≡' else '#4b5563'
        symbol = 'M' if label == 'MetaVLM' else ('R' if label == 'Rand' else '≡')
        ax_matrix.text(col_idx, row_idx, symbol, ha='center', va='center',
                       color=text_color, fontweight='bold', fontsize=9)

ax_matrix.set_xlim(-0.6, 29.6)
ax_matrix.set_ylim(-0.6, 1.6)
ax_matrix.set_xticks(range(30))
ax_matrix.set_xticklabels(benchmarks, fontsize=9, fontweight='bold')
ax_matrix.set_yticks([0, 1])
ax_matrix.set_yticklabels(["CLIP", "BLIP"], fontsize=11, fontweight='bold')
ax_matrix.set_title("(a) Individual Benchmark Hypervolume Winner Matrix ($B_1$–$B_{30}$)", fontsize=12, fontweight='bold', pad=25)
ax_matrix.invert_yaxis()
ax_matrix.spines['top'].set_visible(False)
ax_matrix.spines['right'].set_visible(False)
ax_matrix.spines['left'].set_visible(False)
ax_matrix.spines['bottom'].set_visible(False)
ax_matrix.tick_params(left=False, bottom=False)

legend_elements = [
    Patch(facecolor='#2b5c8f', edgecolor='none', label='MetaVLM Wins (M)'),
    Patch(facecolor='#d1d5db', edgecolor='none', label='No Difference (≡)'),
    Patch(facecolor='#e07a5f', edgecolor='none', label='Rand Wins (R)')
]
ax_matrix.legend(handles=legend_elements, loc='upper center', bbox_to_anchor=(0.5, 1.32),
                 ncol=3, frameon=True, fontsize=10)

# --- Panel 2: Summary Outcome Distribution ---
clip_counts = [clip_best.count('MetaVLM'), clip_best.count('≡'), clip_best.count('Rand')]
blip_counts = [blip_best.count('MetaVLM'), blip_best.count('≡'), blip_best.count('Rand')]

y = np.array([0, 1])
height = 0.38

meta_wins = [clip_counts[0], blip_counts[0]]
no_diff   = [clip_counts[1], blip_counts[1]]
rand_wins = [clip_counts[2], blip_counts[2]]

ax_summary.barh(y, meta_wins, height, label='MetaVLM Wins', color='#2b5c8f')
ax_summary.barh(y, no_diff, height, left=meta_wins, label='No Difference (≡)', color='#d1d5db')
ax_summary.barh(y, rand_wins, height, left=np.array(meta_wins) + np.array(no_diff), label='Rand Wins', color='#e07a5f')

# Inner Bar Annotations
for i in range(2):
    w = meta_wins[i]
    ax_summary.text(w / 2, i, f"{w} ({w/30*100:.1f}%)", ha='center', va='center', color='white', fontweight='bold', fontsize=10)

    w_nd = no_diff[i]
    ax_summary.text(meta_wins[i] + w_nd / 2, i, f"{w_nd} ({w_nd/30*100:.1f}%)", ha='center', va='center', color='#374151', fontweight='bold', fontsize=10)

    w_r = rand_wins[i]
    if w_r > 0:
        ax_summary.text(meta_wins[i] + no_diff[i] + w_r / 2, i, f"{w_r} ({w_r/30*100:.1f}%)", ha='center', va='center', color='white', fontweight='bold', fontsize=10)

ax_summary.set_yticks(y)
ax_summary.set_yticklabels(["CLIP", "BLIP"], fontsize=11, fontweight='bold')
ax_summary.set_xlabel("Number of Benchmark Images (Out of 30)", fontsize=10)
ax_summary.set_xlim(0, 30)
ax_summary.set_title("(b) Summary Outcome Distribution Across 30 Benchmarks", fontsize=12, fontweight='bold', pad=10)
ax_summary.grid(axis='x', linestyle=':', alpha=0.5)
ax_summary.spines['top'].set_visible(False)
ax_summary.spines['right'].set_visible(False)

#plt.suptitle("Table 4: RQ1 - Hypervolume Performance Comparison (MetaVLM vs. Rand)", fontsize=14, fontweight='bold', y=0.98)
plt.subplots_adjust(top=0.88, bottom=0.08, hspace=0.5)
plt.savefig('Table4_RQ1_Performance_Dashboard.png', dpi=300, bbox_inches='tight')
plt.show()
```
